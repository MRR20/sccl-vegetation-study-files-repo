```python
import xarray as xr
import pandas as pd
from pathlib import Path
```

```python
data_folder = Path(r"C:/Users/RUTHVIK REDDY/lulc_project/imd rainfall dataset 2004 to 2025")
```

```python
WEST = 79.2227885552
EAST = 79.7192575950
SOUTH = 18.5522922817
NORTH = 19.1367296520
```

```python
files = sorted(data_folder.glob("*.nc"))

print("Number of files:", len(files))

results = []
```

```python
for file in files:

    print("Processing:", file.name)

    ds = xr.open_dataset(file)

    rain = ds["RAINFALL"]

    study_rain = rain.sel(
        LATITUDE=slice(SOUTH, NORTH),
        LONGITUDE=slice(WEST, EAST)
    )

    daily_mean = study_rain.mean(
        dim=["LATITUDE", "LONGITUDE"]
    )

    annual_rainfall = float(
        daily_mean.sum(dim="TIME").values
    )

    monsoon = daily_mean.sel(
        TIME=daily_mean.TIME.dt.month.isin([6, 7, 8, 9])
    )

    monsoon_rainfall = float(
        monsoon.sum(dim="TIME").values
    )

    year = int(daily_mean.TIME.dt.year.values[0])

    results.append({
        "Year": year,
        "Annual_Rainfall_mm": annual_rainfall,
        "Monsoon_Rainfall_mm": monsoon_rainfall
    })

    ds.close()
```

```python
df = pd.DataFrame(results)

df = df.sort_values("Year").reset_index(drop=True)

print(df)
```

```python
output_file = data_folder / "IMD_Rainfall_Annual_Monsoon.csv"

df.to_csv(output_file, index=False)

print("\nSaved to:")
print(output_file)
```
