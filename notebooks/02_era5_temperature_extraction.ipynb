```python
import xarray as xr
import pandas as pd


grib_file = r"C:/Users/RUTHVIK REDDY/lulc_project/Temperature data ERA5 2004 to 2026/1f24f8f242660c61637162454ad340d4.grib"

ds = xr.open_dataset(
    grib_file,
    engine="cfgrib",
    backend_kwargs={"indexpath": ""}
)


WEST = 79.2227885552
EAST = 79.7192575950
SOUTH = 18.5522922817
NORTH = 19.1367296520


temp = ds["t2m"].sel(
    time=slice("2004-01-01", "2025-12-31")
)


study_temp = temp.sel(
    latitude=slice(NORTH, SOUTH),
    longitude=slice(WEST, EAST)
)


regional_temp = study_temp.mean(
    dim=["latitude", "longitude"]
)


regional_temp_c = regional_temp - 273.15

print(regional_temp_c)
```

```python
annual_temp = regional_temp_c.groupby("time.year").mean()

print(annual_temp)
```

```python
jan_feb_temp = regional_temp_c.sel(
    time=regional_temp_c.time.dt.month.isin([1, 2])
)

jan_feb_temp = jan_feb_temp.groupby("time.year").mean()

print(jan_feb_temp)
```

```python
temperature_df = pd.DataFrame({
    "Year": annual_temp.year.values,
    "Annual_Mean_Temperature_C": annual_temp.values,
    "Jan_Feb_Mean_Temperature_C": jan_feb_temp.values
})

print(temperature_df)
```

```python
temperature_df.to_csv(
    "ERA5_Temperature_2004_2025.csv",
    index=False
)
```
